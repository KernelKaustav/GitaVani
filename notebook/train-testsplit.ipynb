```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
from pathlib import Path
```

```python
df = pd.read_csv("C:/UserData/Kaustav/python projects/GitaVani/dataset/processed/metadata.csv")
df.head()
```

```python
ravdess = df[df["dataset"] == "RAVDESS"].copy()
crema = df[df["dataset"] == "CREMA-D"].copy()
tess = df[df["dataset"] == "TESS"].copy()
```

```python
ravdess["speaker"].unique()
```

```python
np.random.seed(42)
speakers = ravdess["speaker"].unique()
np.random.shuffle(speakers)
len(speakers)
```

```python
train_sp = speakers[:17]
val_sp = speakers[17:20]
test_sp = speakers[20:]
```

```python
rav_train = ravdess[ravdess["speaker"].isin(train_sp)]
rav_val = ravdess[ravdess["speaker"].isin(val_sp)]
rav_test = ravdess[ravdess["speaker"].isin(test_sp)]
```

```python
speakers = crema["speaker"].unique()
np.random.shuffle(speakers)
n = len(speakers)
train_sp = speakers[:int(0.7*n)]
val_sp = speakers[int(0.7*n):int(0.85*n)]
test_sp = speakers[int(0.85*n):]
crema_train = crema[crema["speaker"].isin(train_sp)]
crema_val = crema[crema["speaker"].isin(val_sp)]
crema_test = crema[crema["speaker"].isin(test_sp)]
```

```python
tess_train, temp = train_test_split(
    tess,
    test_size=0.30,
    stratify=tess["emotion"],
    random_state=42
)

tess_val, tess_test = train_test_split(
    temp,
    test_size=0.50,
    stratify=temp["emotion"],
    random_state=42
)
```

```python
train = pd.concat([
    rav_train,
    crema_train,
    tess_train
], ignore_index=True)

val = pd.concat([
    rav_val,
    crema_val,
    tess_val
], ignore_index=True)

test = pd.concat([
    rav_test,
    crema_test,
    tess_test
], ignore_index=True)
```

```python
print(len(train))
print(len(val))
print(len(test))
print(train["emotion"].value_counts())
print(val["emotion"].value_counts())
print(test["emotion"].value_counts())
```

```python
processed = Path("C:/UserData/Kaustav/python projects/GitaVani/dataset/processed")

train.to_csv(processed/"train.csv", index=False)

val.to_csv(processed/"val.csv", index=False)

test.to_csv(processed/"test.csv", index=False)

print("Done!")
```
